This is a **synthetic supervised-learning illustration**, not molecular data. Eight independently generated groups stand in for independent runs. The feature is a random number and the label is whether it exceeds zero plus noise. Train and test groups are separated. Compare an interpretable threshold with a nearest-centroid classifier and a constant-label baseline.

```python
import numpy as np
rng=np.random.default_rng(8)
x=rng.normal(size=(8,80))
y=(x+0.4*rng.normal(size=x.shape)>0).astype(int)
xtrain,ytrain=x[:5].ravel(),y[:5].ravel()
xtest,ytest=x[5:].ravel(),y[5:].ravel()
centres=np.array([xtrain[ytrain==c].mean() for c in (0,1)])
pred=np.argmin((xtest[:,None]-centres[None,:])**2,axis=1)
constant=np.full_like(ytest,int(ytrain.mean()>=0.5))
threshold=(xtest>0).astype(int)
for label,p in [("constant",constant),("threshold",threshold),("nearest centroid",pred)]:
    print(label,"held-out accuracy",np.mean(p==ytest))
print("Test positive fraction",ytest.mean())
```
