Population: repeated fair binary trials. X is a fair bit and Y is either an independent fair bit or a copy of X. The joint law has respectively four equally likely pairs or two diagonal pairs. The product law draws X and Y independently from their marginals. Exact MI is respectively 0 or 1 bit. We estimate from many pairs, never from one outcome.

```python
import numpy as np
def mi(table):
    p=np.asarray(table,dtype=float);p=p/p.sum()
    q=p.sum(axis=1,keepdims=True)*p.sum(axis=0,keepdims=True)
    mask=p>0
    return float(np.sum(p[mask]*np.log2(p[mask]/q[mask])))
assert abs(mi([[1,1],[1,1]]))<1e-12
assert abs(mi([[1,0],[0,1]])-1)<1e-12
rng=np.random.default_rng(12)
x=rng.integers(0,2,1000); independent=rng.integers(0,2,1000)
counts=np.zeros((2,2),int);np.add.at(counts,(x,independent),1)
print("Finite-sample independent estimate:",mi(counts),"bits; exact population value: 0")
```

Why need not the finite estimate be exactly zero? What changes if we use fewer trials? This exercise is not a measurement of thermodynamic entropy.
