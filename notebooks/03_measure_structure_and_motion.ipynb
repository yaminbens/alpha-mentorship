Choose the same particle identities at two times. For each particle define its k nearest neighbours; persistence is the retained fraction averaged over particles. Exact distance ties use index order. Periodic distances require the same box in both frames. This is a dynamical measure that depends on the time lag.

```python
import numpy as np
import matplotlib.pyplot as plt
def neighbours(frame, k=4, box=None):
    frame = np.asarray(frame, dtype=float)
    if frame.ndim != 2 or not 1 <= k < len(frame):
        raise ValueError("Need positions (particles, dimensions) and 1 <= k < N")
    delta = frame[:, None, :] - frame[None, :, :]
    if box is not None:
        delta -= box * np.round(delta / box)
    distance2 = np.sum(delta**2, axis=-1)
    np.fill_diagonal(distance2, np.inf)
    return np.argsort(distance2, axis=1, kind="stable")[:, :k]

def persistence(first, later, k=4, box=None):
    a,b = neighbours(first,k,box), neighbours(later,k,box)
    return np.mean([len(set(x)&set(y))/k for x,y in zip(a,b)])
```

```python
rng=np.random.default_rng(7)
first=rng.uniform(0,10,(20,2))
translated=first+np.array([2,3])
print("Translation control:",persistence(first,translated,k=4))
assert persistence(first,translated,k=4)==1.0
later=first+rng.normal(0,1,first.shape)
print("Synthetic displaced configuration:",persistence(first,later,k=4))
```
