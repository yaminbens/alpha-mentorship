```python
import io
import pandas as pd
import matplotlib.pyplot as plt
using_example = True
csv_text = """condition,repeat,T_reduced,U_per_particle
low,1,0.1,-2.2
middle,1,0.3,-1.8
high,1,0.6,-1.2
middle,2,0.32,-1.75
"""
table = pd.read_csv(io.StringIO(csv_text))
required = ["T_reduced", "U_per_particle"]
for column in required:
    table[column] = pd.to_numeric(table[column], errors="raise")
print("INVENTED FORMAT EXAMPLE" if using_example else "STUDENT OBSERVATIONS — verify source and units")
print(table.to_string(index=False))
```

```python
plt.scatter(table.T_reduced, table.U_per_particle)
for _,row in table.iterrows():
    plt.annotate(str(row["condition"])+" / "+str(row["repeat"]), (row.T_reduced,row.U_per_particle))
plt.xlabel("Measured k_B T / epsilon"); plt.ylabel("Potential energy / (N epsilon)")
plt.title("Invented example" if using_example else "My observations"); plt.show()
table.to_csv("recorded_observations.csv",index=False)
```
