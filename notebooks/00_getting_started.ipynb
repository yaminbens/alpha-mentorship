```python
message = "המחברת האישית שלי"
print(message)
scale = 1.0
```

## תורכם לשנות ולהוסיף
הריצו את תא הגרף הבא. שנו את `scale` ל־2, הריצו שוב את תא המשתנה ואחריו את תא הגרף. הוסיפו לתא הראשון שורה שמדפיסה את הערך של `scale`: המבנה הוא `print(שם_המשתנה)`.

הנקודות כאן מומצאות לצורך תרגול. ציר `time` הוא זמן לדוגמה, וציר `value` הוא גודל לדוגמה; אלה אינם נתוני סימולציה.

</div>

```python
import numpy as np
import matplotlib.pyplot as plt
time = np.array([0, 1, 2, 3, 4])
values = scale * np.array([1.0, 1.4, 1.2, 1.6, 1.5])
plt.plot(time, values, "o-")
plt.xlabel("Example time"); plt.ylabel("Example value"); plt.show()
```
