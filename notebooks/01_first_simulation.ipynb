```python
# Fetch three MIT-licensed teaching modules and their license from a pinned commit.
# The content hashes verify the exact source used by this course.
import urllib.request, hashlib, sys, tempfile
from pathlib import Path
MODULES = {'mathTools.py': '17e6c29a9348b42c1a68c04caba66762d89b483110013b6a6facf5a3792012cc', 'classes.py': '90c53b632780b20d8b6b0e9078cc24d9abcf7e5480cd4ebdd5b8af37fb8f0a36', 'dynamics.py': '0db62c33a9c5fd68ef6dd8c251abe5b44810317127cf6c29e283868b4d32997d', 'LICENSE': '36de33ca97bf89a872c7e00578b7b1e71514276cba6f189ca003bf621b2db622'}
module_dir = Path(tempfile.gettempdir()) / "alpha_lj_pinned"
module_dir.mkdir(exist_ok=True)
for filename, expected in MODULES.items():
    path = module_dir / filename
    if not path.exists() or hashlib.sha256(path.read_bytes()).hexdigest() != expected:
        content = urllib.request.urlopen("https://raw.githubusercontent.com/ggrazioli/LennardJonesFluidLab/9f3f932bb3ffdf47bc0de23378aba196bdc9452b/" + filename).read()
        if hashlib.sha256(content).hexdigest() != expected:
            raise ValueError("Source integrity mismatch: " + filename)
        path.write_bytes(content)
sys.path.insert(0, str(module_dir))
import numpy as np
import matplotlib.pyplot as plt
from classes import Ensemble
from dynamics import takeOneStep
```

```python
# Small CPU teaching example, not a melting experiment.
np.random.seed(42)
system = Ensemble(nMol=16, sideLen=8.0, dim=2, temperature=0.3, periodic=True)
system.UpdateAccelerationLJ()  # initialize forces before the first integration step
positions, values = [], []
dt = 0.002
for step in range(201):
    if step % 5 == 0:
        positions.append(np.array([m.r.copy() for m in system.moList]))
        kinetic = system.getKE()
        values.append([step*dt, 2*kinetic/(system.dim*system.nMol), system.getPE()/system.nMol,
                       (kinetic+system.getPE())/system.nMol])
    if step < 200:
        takeOneStep(system, dt=dt)
positions = np.asarray(positions)
values = np.asarray(values)
print("Samples:", len(values), "simulation time:", values[-1,0], "reduced time units")
print("Energy per-particle range:", np.ptp(values[:,3]), "epsilon")
plt.plot(values[:,0], values[:,1]); plt.xlabel("Time / reduced unit"); plt.ylabel("k_B T / epsilon"); plt.show()
```

```python
import json
np.savez("alpha_trajectory.npz", positions=positions, values=values, L=system.sideLen, dt=dt, sample_every=5)
np.savetxt("alpha_observations.csv", values, delimiter=",", header="time,T_reduced,U_per_particle,E_per_particle", comments="")
print("Created alpha_trajectory.npz and alpha_observations.csv — download these before the runtime ends.")
```
